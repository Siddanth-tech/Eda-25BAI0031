```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
```

# 1. Create a dummy dataset with an outlier

```python
data = {
'Age': [25, 30, 45, 35, 22],
'Salary': [50000, 64000, 120000, 85000, 1200000] # 1,200,000 is an outlier
}
df = pd.DataFrame(data)
print("Original Data:\n", df, "\n")
```

# 2. Min-Max Scaling (Normalization)
# Rescales features to a fixed range, typically 0 to 1.

```python
min_max_scaler = MinMaxScaler()
df_min_max = pd.DataFrame(
min_max_scaler.fit_transform(df),
columns=df.columns
)
print("Min-Max Scaled Data:\n", df_min_max, "\n")
```

 3. Z-score Standardization
# Centers data around mean 0 with a standard dev of 1

```python
standard_scaler = StandardScaler()
df_standard = pd.DataFrame(
standard_scaler.fit_transform(df),
columns=df.columns
)
print("Standardized Data:\n", df_standard, "\n")
```
